# Trennung von Trainings- und Validierungsdaten
Auch bei der Voruntersuchung gilt die Regel, dass die später zum Test bzw. zur Validierung verwendeten Daten nicht berücksichtigt werden dürfen, daher ist zunächst eine Trennung der Daten erforderlich. Hierzu kann auf existierende Funktionen der Bibliothek sklearn zur Trennung der Daten zurückgegriffen werden.

Hierbei wird eine Zufallsauswahl getroffen und der Datensatz durch Setzen von test_size=0.2 zu 80% - 20% aufgeteilt. Mit Definition eines random_state wird sichergestellt, dass jedes mal die identische Zufallsauswahl erfolgt, damit bei einer wiederholten Programmausfuehrung nicht jeweils andere Daten in den Testdatensatz aufgenommen und damit bereits vorab beruecksichtigt werden.

```python
from sklearn.model_selection import train_test_split
train, val = train_test_split(data, test_size = 0.2, random_state=42)
```

```python
# Umfang der Datensaetze anzeigen
len(train), len(val)
```

```python
# Kopie des Trainingsdatensatzes anlegen, so, dass die Tabellen auch modifiziert werden koennen
data = train.copy()
```

```python
# Begrenzung der Auswahl auf die Sensorspalten ueber Doppelpunkt-Operator, der aber nur mit .loc funktioniert
data.hist(bins=50, figsize=(20, 15))
plt.show()
```

## Visualisierung der Klassenzuordnung im Scatterplot
Erstellung einer neuen Spalte "class_num" mit einem numerischen Wert je Klasse und Visualierung als farblich codierter Scatterplot


```python
# Textbezeichnung der Klassen in numerische Werte ueberfuehren
data['class_num'], data_categories = data['class'].factorize()
for i, cat in enumerate(data_categories):
    print("Klasse %d: %s" %(i, cat))
```

```python
data.head()
```

```python
# Analyse der Zuordnung zu Klassen
data.plot(kind="scatter", x='y1', y='z1', alpha=0.1, c='class_num', cmap=plt.get_cmap('jet'))
plt.show()
```

```python
# Analyse der Zuordnung zu Klassen
data.plot(kind="scatter", x='y2', y='z2', alpha=0.1, c='class_num', cmap=plt.get_cmap('jet'))
plt.show()
```
